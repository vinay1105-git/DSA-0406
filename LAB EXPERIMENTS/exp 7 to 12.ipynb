```python
import pandas as pd

order_data = pd.DataFrame({
    'customer_id': [101, 102, 101, 103, 102],
    'order_date': ['2026-01-10', '2026-01-12', '2026-01-15', '2026-01-20', '2026-01-25'],
    'product_name': ['Laptop', 'Phone', 'Laptop', 'Tablet', 'Phone'],
    'order_quantity': [2, 3, 4, 1, 5]
})

order_data['order_date'] = pd.to_datetime(order_data['order_date'])

print("Total Orders by Each Customer:")
print(order_data.groupby('customer_id').size())

print("\nAverage Order Quantity for Each Product:")
print(order_data.groupby('product_name')['order_quantity'].mean())

print("\nEarliest Order Date:")
print(order_data['order_date'].min())

print("\nLatest Order Date:")
print(order_data['order_date'].max())
```

```python
import pandas as pd

sales_data = pd.DataFrame({
    'product': ['Laptop', 'Phone', 'Laptop', 'Tablet', 'Phone', 'Mouse'],
    'quantity': [10, 20, 15, 8, 25, 12],
    'date': ['2026-09-05', '2026-09-10', '2026-09-15',
             '2026-09-20', '2026-09-25', '2026-09-28']
})

sales_data['date'] = pd.to_datetime(sales_data['date'])

today = pd.Timestamp.today().normalize()
start_date = today.replace(day=1) - pd.DateOffset(months=1)
end_date = today.replace(day=1)

monthly_sales = sales_data[
    (sales_data['date'] >= start_date) &
    (sales_data['date'] < end_date)
]

top_5 = monthly_sales.groupby('product')['quantity'].sum().nlargest(5)

print("Top 5 Best-Selling Products:")
print(top_5)
```

```python
import pandas as pd

property_data = pd.DataFrame({
    'property_id': [1, 2, 3, 4, 5],
    'location': ['Chennai', 'Bangalore', 'Chennai', 'Hyderabad', 'Bangalore'],
    'bedrooms': [3, 5, 2, 6, 4],
    'area': [1200, 2000, 1500, 2500, 1800],
    'listing_price': [500000, 800000, 600000, 900000, 700000]
})

print("Average Listing Price in Each Location:")
print(property_data.groupby('location')['listing_price'].mean())

print("\nNumber of Properties with More Than Four Bedrooms:")
print((property_data['bedrooms'] > 4).sum())

print("\nProperty with the Largest Area:")
print(property_data.loc[property_data['area'].idxmax()])
```

```python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [1000, 1500, 1200, 1800, 2000, 2500]

plt.figure(figsize=(8, 4))
plt.plot(months, sales, marker='o')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.title('Monthly Sales - Line Plot')
plt.show()

plt.figure(figsize=(8, 4))
plt.bar(months, sales)
plt.xlabel('Month')
plt.ylabel('Sales')
plt.title('Monthly Sales - Bar Plot')
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

temperature = [22, 24, 28, 32, 35, 34, 30, 29, 28, 27, 25, 22]
rainfall = [20, 15, 25, 40, 60, 120, 180, 160, 140, 100, 50, 25]

plt.figure(figsize=(8, 4))
plt.plot(months, temperature, marker='o')
plt.xlabel('Month')
plt.ylabel('Temperature (°C)')
plt.title('Monthly Temperature - Line Plot')
plt.show()

plt.figure(figsize=(8, 4))
plt.scatter(months, rainfall)
plt.xlabel('Month')
plt.ylabel('Rainfall (mm)')
plt.title('Monthly Rainfall - Scatter Plot')
plt.show()
```
