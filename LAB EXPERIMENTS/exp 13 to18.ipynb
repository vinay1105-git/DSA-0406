```python
from collections import Counter
import re

text = """Python is easy to learn. Python is used
for data science and data analysis."""

words = re.findall(r'\b\w+\b', text.lower())

frequency = Counter(words)

print("Word Frequency Distribution:")
for word, count in frequency.items():
    print(word, ":", count)
```

```python
from collections import Counter
import re

reviews = [
    "Good product and good quality",
    "Excellent product",
    "Good quality and fast delivery",
    "Excellent service"
]

words = []

for review in reviews:
    words.extend(re.findall(r'\b\w+\b', review.lower()))

frequency = Counter(words)

print("Word Frequency Distribution:")
for word, count in frequency.items():
    print(word, ":", count)
```

```python
import pandas as pd
import re
import matplotlib.pyplot as plt
from collections import Counter

# Create sample dataset
data = {
    'feedback': [
        'The product is very good',
        'Good product and excellent quality',
        'The delivery is very fast',
        'Excellent product and good service',
        'The product quality is good',
        'Fast delivery and excellent service'
    ]
}

df = pd.DataFrame(data)

# Save dataset as CSV
df.to_csv('data.csv', index=False)

# Read CSV file
df = pd.read_csv('data.csv')

# Stop words
stop_words = {'the', 'and', 'is', 'a', 'an', 'to', 'of', 'in', 'for'}

words = []

# Preprocessing
for feedback in df['feedback'].dropna():
    text = re.sub(r'[^\w\s]', '', str(feedback).lower())

    for word in text.split():
        if word not in stop_words:
            words.append(word)

# Word frequency
frequency = Counter(words)

# Get top N words
n = int(input("Enter top N words: "))
top_words = frequency.most_common(n)

# Display results
print("\nTop N Frequent Words:")
for word, count in top_words:
    print(word, ":", count)

# Bar graph
if top_words:
    w, c = zip(*top_words)

    plt.bar(w, c)
    plt.xlabel("Words")
    plt.ylabel("Frequency")
    plt.title("Top N Frequent Words")
    plt.xticks(rotation=45)
    plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats

data = {
    'Age': [23, 23, 27, 27, 39, 41, 47, 49, 50,
            52, 54, 54, 56, 57, 58, 58, 60, 61],

    'Fat': [9.5, 26.5, 7.8, 17.8, 31.4, 25.9, 27.4, 27.2, 31.2,
            34.6, 42.5, 28.8, 33.4, 30.2, 34.1, 32.9, 41.2, 35.7]
}

df = pd.DataFrame(data)

print("Statistical Measures:")
print(df.describe().loc[['mean', '50%', 'std']])

# Boxplot
plt.figure(figsize=(8, 5))
sns.boxplot(data=df)
plt.title("Boxplot of Age and Body Fat")
plt.show()

# Scatter Plot
plt.figure(figsize=(8, 5))
plt.scatter(df['Age'], df['Fat'])
plt.xlabel("Age")
plt.ylabel("Body Fat (%)")
plt.title("Age vs Body Fat")
plt.show()

# Q-Q Plot for Age
stats.probplot(df['Age'], dist="norm", plot=plt)
plt.title("Q-Q Plot of Age")
plt.show()

# Q-Q Plot for Body Fat
stats.probplot(df['Fat'], dist="norm", plot=plt)
plt.title("Q-Q Plot of Body Fat")
plt.show()
```
